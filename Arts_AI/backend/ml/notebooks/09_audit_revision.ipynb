# Audit revision summary
Loads preserved original metrics and revised experiment outputs. Does not delete prior runs.

```python
import json
from pathlib import Path
from arts_ai.common.paths import get_paths

p = get_paths()
audit = p.reports / 'audit'
print('Audit files:', sorted(x.name for x in audit.glob('*')) if audit.exists() else None)
for name in ['c4_topic_metrics_REVISED.json','c2_ranking_REVISED.json','c3_uci_metrics_REVISED.json']:
    f = p.reports / name
    print('\n===', name, '===')
    if f.exists():
        data = json.loads(f.read_text(encoding='utf-8'))
        print(json.dumps(data, indent=2, default=str)[:2000])
    else:
        print('missing')
print('\nLatest topic run:', (p.artifact('c4')/'LATEST_TOPIC_RUN.txt').read_text().strip())
print('Latest rank run:', (p.artifact('c2')/'LATEST_RANK_RUN.txt').read_text().strip())
```
